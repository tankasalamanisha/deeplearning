Adding problem — sequence of (value, marker) pairs, output the sum of the two marked values. Sweep T = 100, 200, 500, 1000. Random-guess MSE is ~0.167; anything above that means your model learned nothing. This is the canonical long-range credit assignment test.<br><br>

Input is T×2. Channel 1 is i.i.d. uniform on [0,1]. Channel 2 is all zeros with exactly two ones: one at a uniform random position in the first half, one in the second half. Target is the scalar sum of the two marked values. MSE loss, single output at the last timestep. The variance of a sum of two U[0,1] draws is 1/6, so a model that always predicts 1.0 gets MSE ≈ 0.167 — that's your "learned nothing" floor, and models sit on it for a long time before dropping.

## What is the Adding Problem?
<br>

Input Structure: The model receives two parallel sequences of length T (where T can be large, such as 100, 200, or 500+).
First row: Random real numbers uniformly sampled between 0 and 1.<br><br>
Second row: A marker sequence containing all zeros except for exactly two entries marked with a 1—one in the first half of the sequence and one in the second half. <br><br>

Target Output: A single scalar value equal to the sum of the two numbers from the first row that correspond to the positions of the two 1 markers in the second row.
